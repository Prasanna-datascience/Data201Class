# Simple Scatter Plots

Another commonly used plot type is the simple scatter plot, a close cousin of the line plot.
Instead of points being joined by line segments, here the points are represented individually with a dot, circle, or other shape.
We’ll start by setting up the notebook for plotting and importing the packages we will use:

```python

# Import Matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Data
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

# Create scatter plot
plt.scatter(study_hours, exam_scores, color='blue', marker='o')

# Add title and axis labels
plt.title('Study Time vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')

# Display plot
plt.grid(True)
plt.show()

# Conclusion
# There is a strong relationship between study time and exam score.
#As the number of hours spent studying increases, exam scores also increases.
#The points on the scatter plot follow an upward trend, suggesting that students who study more tend to achieve higher exam scores.
```
